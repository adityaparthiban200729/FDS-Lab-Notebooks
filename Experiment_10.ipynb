## Array Functions in Data Science (NumPy)

**Aim:** To understand array functions in Data science using the NumPy library.

```python
import numpy as np

np.random.seed(42)
array = np.random.randint(1, 100, 9)  # randomly generate 9 numbers between 1 to 100
array
```

```python
np.sqrt(array)
```

```python
array.ndim
```

```python
new_array = array.reshape(3, 3)
new_array
```

```python
new_array.ndim
```

```python
new_array.ravel()
```

```python
newm = new_array.reshape(3, 3)
newm
```

```python
newm[2, 1:3]
```

```python
newm[1:2, 1:3]
```

```python
new_array[0:3, 0:0]
```

```python
new_array[0:2, 0:1]
```

```python
new_array[0:3, 0:1]
```
