## Outlier Detection in a Given Data Set

**Aim:** To understand the procedure to identify the outliers in a given data set.

```python
# sample calculation for low range (lr), upper range (ur), percentile
import numpy as np

array = np.array([27, 50, 44, 6, 58, 61, 23, 86, 67, 20, 75, 7, 79, 61, 90, 54])
array
```

```python
array.mean()
```

```python
np.percentile(array, 25)
```

```python
np.percentile(array, 50)
```

```python
np.percentile(array, 75)
```

```python
np.percentile(array, 100)
```

```python
# outliers detection
def outDetection(array):
    sorted(array)
    Q1, Q3 = np.percentile(array, [25, 75])
    IQR = Q3 - Q1
    lr = Q1 - (1.5 * IQR)
    ur = Q3 + (1.5 * IQR)
    return lr, ur

lr, ur = outDetection(array)
lr, ur
```

```python
%matplotlib inline
import seaborn as sns
import matplotlib.pyplot as plt

sns.displot(array)
plt.show()
```

```python
new_array = array[(array > lr) & (array < ur)]
new_array
```

```python
sns.displot(new_array)
plt.show()
```

```python
lr1, ur1 = outDetection(new_array)
lr1, ur1
```

```python
final_array = new_array[(new_array > lr1) & (new_array < ur1)]
final_array
```

```python
sns.displot(final_array)
plt.show()
```
