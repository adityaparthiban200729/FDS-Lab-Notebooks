## Data Preprocessing in Data Science

**Aim:** To understand the importance of data preprocessing in data science.

```python
import numpy as np
import pandas as pd

data = {
    'Country': ['France', 'Spain', 'Germany', 'Spain', 'Germany',
                'France', 'Spain', 'France', np.nan, 'France'],
    'Age': [44, 27, 30, 38, 40, 35, np.nan, 48, 50, 37],
    'Salary': [72000, 48000, 54000, 61000, np.nan, 58000, 52000,
               79000, 83000, 67000],
    'Purchased': ['No', 'Yes', 'No', 'No', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}
pd.DataFrame(data).to_csv('pre-process_datasample.csv', index=False)

df = pd.read_csv('pre-process_datasample.csv')
df
```

```python
df.info()
```

```python
df.Country.mode()
```

```python
df.Country.mode()[0]
```

```python
type(df.Country.mode())
```

```python
df['Country'] = df['Country'].fillna(df['Country'].mode()[0])
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Salary'] = df['Salary'].fillna(round(df['Salary'].mean()))
df
```

```python
pd.get_dummies(df.Country)
```

```python
updated_dataset = pd.concat([pd.get_dummies(df.Country), df.iloc[:, [1, 2, 3]]], axis=1)
updated_dataset
```

```python
df.info()
```

```python
updated_dataset['Purchased'] = updated_dataset['Purchased'].replace(['No', 'Yes'], [0, 1])
updated_dataset
```
