## Handling Missing Values and Inappropriate Data in a Data Set

**Aim:** To demonstrate an experiment to handle missing data and inappropriate data in a data set using the Python pandas library for data preprocessing.

```python
# Upload Hotel.csv and convert it into dataFrame
import numpy as np
import pandas as pd

data = {
    'CustomerID': [1, 2, 3, 4, 5, 6, 7, 8, 9, 9, 10],
    'Age_Group': ['20-25', '30-35', '25-30', '20-25', '35+', '35+', '35+',
                  '20-25', '25-30', '25-30', '30-35'],
    'Rating(1-5)': [4, 5, 6, -1, 3, 3, 4, 7, 2, 2, 5],
    'Hotel': ['Ibis', 'LemonTree', 'RedFox', 'LemonTree', 'Ibis', 'Ibys',
              'RedFox', 'LemonTree', 'Ibis', 'Ibis', 'RedFox'],
    'FoodPreference': ['veg', 'Non-Veg', 'Veg', 'Veg', 'Vegetarian', 'Non-Veg',
                       'Vegetarian', 'Veg', 'Non-Veg', 'Non-Veg', 'non-Veg'],
    'Bill': [1300, 2000, 1322, 1234, 989, 1909, 1000, 2999, 3456, 3456, -6755],
    'NoOfPax': [2, 3, 2, 2, 2, 2, -1, -10, 3, 3, 4],
    'EstimatedSalary': [40000, 59000, 30000, 120000, 45000, 122220, 21122,
                        345673, -99999, -99999, 87777],
    'Age_Group.1': ['20-25', '30-35', '25-30', '20-25', '35+', '35+', '35+',
                    '20-25', '25-30', '25-30', '30-35']
}
df = pd.DataFrame(data)
df.to_csv('Hotel_Dataset.csv', index=False)
print("Hotel_Dataset.csv created successfully!")
```

```python
df = pd.read_csv('Hotel_Dataset.csv')
df
```

```python
# From the dataframe identify the duplicate row (i.e row 9)
# The duplicated() method returns a Series with True and False values
# that describe which rows in the DataFrame are duplicated and not.
df.duplicated()
```

```python
# The info() method prints information about the DataFrame.
df.info()
```

```python
# The drop_duplicates() method removes duplicate rows.
df.drop_duplicates(inplace=True)
df
```

```python
# While removing duplicate record row index also removed.
# The len() function returns the number of rows.
print(len(df))

# Reset the index
df.reset_index(drop=True, inplace=True)
df
```

```python
# Axis refers to the dimensions of a DataFrame (index and columns).
# Use axis=1 to apply functions column-wise across columns.
# Drop the repeated Age_Group column.
df.drop(['Age_Group.1'], axis=1, inplace=True)
df
```

```python
# The function .loc is typically used for label indexing and can access
# multiple columns. Replace inappropriate negative values with NaN.
df.loc[df.CustomerID < 0, 'CustomerID'] = np.nan
df.loc[df.Bill < 0, 'Bill'] = np.nan
df.loc[df.EstimatedSalary < 0, 'EstimatedSalary'] = np.nan
df
```

```python
# NoOfPax (number of people) must be between 1 and 20
df.loc[(df['NoOfPax'] < 1) | (df['NoOfPax'] > 20), 'NoOfPax'] = np.nan
df
```

```python
# Fix the spelling mistake 'Ibys' -> 'Ibis'
df.Hotel.replace(['Ibys'], 'Ibis', inplace=True)
df.Hotel.unique()
```

```python
df.FoodPreference.unique()
```

```python
# Standardize the FoodPreference categories
df.FoodPreference.replace(['Vegetarian', 'veg'], 'Veg', inplace=True)
df.FoodPreference.replace(['non-Veg'], 'Non-Veg', inplace=True)
df.FoodPreference.unique()
```

```python
# Fillna is a Pandas function to fill the NA/NaN values.
# Numerical continuous data -> replace missing (NaN) value by mean.
# Numerical discrete data -> replace missing (NaN) value by median.
# Categorical data -> replace missing (NaN) value by mode.
df['EstimatedSalary'] = df['EstimatedSalary'].fillna(round(df['EstimatedSalary'].mean()))
df['NoOfPax'] = df['NoOfPax'].fillna(round(df['NoOfPax'].median()))
df['Rating(1-5)'] = df['Rating(1-5)'].fillna(round(df['Rating(1-5)'].median()))
df['Bill'] = df['Bill'].fillna(round(df['Bill'].mean()))
df
```
