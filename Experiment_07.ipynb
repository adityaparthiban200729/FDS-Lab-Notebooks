## Random Sampling and Sampling Distribution

**Aim:** To explore random sampling from a population and understand the concept of sampling distribution using Python.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Step 1: Generate a population (e.g., normal distribution)
population_mean = 50
population_std = 10
population_size = 100000
population = np.random.normal(population_mean, population_std, population_size)
print("Population size:", len(population))
print("Population mean:", round(np.mean(population), 2))
print("Population standard deviation:", round(np.std(population), 2))
```

```python
# Step 2: Random sampling
sample_sizes = [30, 50, 100]  # different sample sizes to consider
num_samples = 1000  # number of samples for each sample size

sample_means = {}
for size in sample_sizes:
    sample_means[size] = []
    for _ in range(num_samples):
        sample = np.random.choice(population, size=size, replace=False)
        sample_means[size].append(np.mean(sample))

print("First 5 sample means for sample size 30:", np.round(sample_means[30][:5], 2))
```

```python
# Step 3: Plotting sampling distributions
plt.figure(figsize=(12, 8))

for i, size in enumerate(sample_sizes):
    plt.subplot(len(sample_sizes), 1, i + 1)
    plt.hist(sample_means[size], bins=30, alpha=0.7, label=f'Sample Size {size}')
    plt.axvline(np.mean(population), color='red', linestyle='dashed',
                linewidth=1.5, label='Population Mean')
    plt.title(f'Sampling Distribution (Sample Size {size})')
    plt.xlabel('Sample Mean')
    plt.ylabel('Frequency')
    plt.legend()

plt.tight_layout()
plt.show()
```
