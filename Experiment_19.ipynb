```python
features = df.iloc[:, [2, 3]].values  # Age and EstimatedSalary
label = df.iloc[:, 4].values          # Purchased
print(features[:10])
print(label[:10])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

# Try a few random states and note where the test score beats the train score
for i in range(1, 21):
    x_train, x_test, y_train, y_test = train_test_split(features, label,
                                                        test_size=0.2,
                                                        random_state=i)
    model = LogisticRegression()
    model.fit(x_train, y_train)
    train_score = model.score(x_train, y_train)
    test_score = model.score(x_test, y_test)
    if test_score > train_score:
        print("Test {} Train {} Random State {}".format(test_score, train_score, i))
```

```python
x_train, x_test, y_train, y_test = train_test_split(features, label,
                                                    test_size=0.2,
                                                    random_state=4)
finalModel = LogisticRegression()
finalModel.fit(x_train, y_train)
```

```python
print("Training accuracy:", finalModel.score(x_train, y_train))
print("Testing accuracy:", finalModel.score(x_test, y_test))
```

```python
from sklearn.metrics import classification_report
print(classification_report(label, finalModel.predict(features)))
```
